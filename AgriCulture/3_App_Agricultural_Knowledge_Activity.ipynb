## Step 5 — Prepare the questions for searching

We will use a simple method called **TF-IDF**.

You do not need to understand the mathematics for this exercise.

The important idea is:

> Python converts the questions into a form that allows it to compare how similar their words are.


```python
questions = data["Question"].fillna("")

vectorizer = TfidfVectorizer(stop_words="english")

question_vectors = vectorizer.fit_transform(questions)

print("The knowledge base is ready for searching.")
```

## Step 6 — Create a simple search function

The function below will:

1. Take your question.
2. Compare it with all questions in the knowledge base.
3. Find the closest question.
4. Return its answer.

Run the cell once to create the function.


```python
def search_knowledge_base(user_question):

    user_vector = vectorizer.transform([user_question])

    similarities = cosine_similarity(
        user_vector,
        question_vectors
    )[0]

    best_match = similarities.argmax()

    matched_question = data.iloc[best_match]["Question"]
    matched_answer = data.iloc[best_match]["Answer"]
    score = similarities[best_match]

    print("YOUR QUESTION")
    print(user_question)

    print("\nCLOSEST QUESTION IN THE KNOWLEDGE BASE")
    print(matched_question)

    print("\nRETRIEVED ANSWER")
    print(matched_answer)

    print("\nSIMILARITY SCORE")
    print(round(score, 3))
```

### Step 10 — The Generation Stage (Completing the RAG)

Instead of an API-based model, we can use an open-source model like **Qwen 2.5 3B**. This model is highly capable and small enough to run on a Colab GPU using quantization.

```python
!pip install -q -U transformers accelerate bitsandbytes

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

# 1. Load Qwen 2.5 3B Instruct in 4-bit to save memory
model_name = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model_qwen = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    quantization_config=None, # For T4 GPU, we rely on high-performance loading
    torch_dtype=torch.bfloat16
)

qwen_pipeline = pipeline(
    "text-generation",
    model=model_qwen,
    tokenizer=tokenizer,
)

def generate_qwen_rag_response(user_query):
    # A. Retrieval Stage
    matched_q, matched_a, score = advanced_search(user_query)

    # B. Generation Stage for Qwen
    messages = [
        {"role": "system", "content": "You are an expert Agricultural Advisor. Use the provided context to answer the farmer's question politely."},
        {"role": "user", "content": f"Question: {user_query}\nContext: {matched_a}"}
    ]

    # Format the prompt using Qwen's template
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    outputs = qwen_pipeline(
        text,
        max_new_tokens=256,
        do_sample=True,
        temperature=0.7,
        top_p=0.9
    )

    response_text = outputs[0]["generated_text"][len(text):]

    print(f"--- QWEN RAG SYSTEM OUTPUT ---")
    print(f"USER QUERY: {user_query}")
    print(f"RETRIEVED FACT: {matched_a}")
    print(f"\nFINAL ADVISORY:\n{response_text}")
```

### Step 12 — Deploying the Agricultural Advisor Website

We will now use **Gradio** to build the user interface. This creates a text box for the farmer and separate output boxes for the database retrieval and the AI's final advice.

```python
import gradio as gr
import torch
from transformers import pipeline
import scipy.io.wavfile as wavfile
import numpy as np

# 1. Initialize the TTS Pipeline
tts_pipe = pipeline("text-to-speech", model="facebook/mms-tts-eng", device="cuda" if torch.cuda.is_available() else "cpu")

def rag_website_interface(user_query):
    # A. Retrieval Stage
    matched_q, matched_a, score = advanced_search(user_query)

    # B. Generation Stage (Qwen)
    messages = [
        {"role": "system", "content": "You are an expert Agricultural Advisor. Use the provided context to answer the farmer's question politely."},
        {"role": "user", "content": f"Question: {user_query}\nContext: {matched_a}"}
    ]

    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    outputs = qwen_pipeline(text, max_new_tokens=256, do_sample=True, temperature=0.7, top_p=0.9)
    response_text = outputs[0]["generated_text"][len(text):]

    # C. TTS Stage
    speech = tts_pipe(response_text)
    # Gradio audio expects (sample_rate, numpy_array)
    audio_output = (speech["sampling_rate"], speech["audio"].squeeze())

    return matched_a, response_text, audio_output

# Define the Updated Web UI
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🌽 Mkulima AI: Agricultural Knowledge Assistant (with Audio)")
    gr.Markdown("Ask any question about maize farming to receive text and audio advice.")

    with gr.Row():
        with gr.Column():
            query_input = gr.Textbox(label="Your Question", placeholder="e.g., How do I protect corn from pests?")
            submit_btn = gr.Button("Ask Assistant", variant="primary")

    with gr.Row():
        with gr.Column():
            retrieval_output = gr.Textbox(label="Step 1: Retrieved Fact (from Database)", interactive=False)
        with gr.Column():
            advisory_output = gr.Textbox(label="Step 2: AI Final Advisory", interactive=False)
            audio_output = gr.Audio(label="Listen to Advisory", type="numpy")

    submit_btn.click(fn=rag_website_interface, inputs=query_input, outputs=[retrieval_output, advisory_output, audio_output])

# Launch the app
demo.launch(share=True, debug=False)
```
